## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can claim: this work is observational and decision-support only. I can
say that certain measurable signals — position, content age, impressions,
engagement — are associated with whether a page's impressions are trending
up or down, and that a model trained on historical client data can rank
pages by estimated risk better than a hand-written rule, based on holdout
precision. Results are directional and useful for prioritizing which pages a
content team reviews first, not a guarantee about any individual page's future.

What I can't claim: I'm not predicting Google's ranking algorithm, and I'm
not proving causation. The starter label, `trend_direction == "down"`, is a
proxy label, not a future outcome — it's a bucket calculated from the current
window (>20% drop in impressions over the most recent 30 days vs. the prior
30), not something measured after a decision point. That means it can't
distinguish real decline from look-alikes: consolidation (a sibling page
absorbing the traffic), seasonality, SERP/AI-driven click loss, or plain
noise from a single volatile month. 54.2% of pages carry this label — close
to a coin flip — so I should describe pages as "flagged by this proxy rule,"
not "confirmed declining." A stronger version of this work would use a
future-window label instead (prior 90 days of features predicting the next
30 days), which is the direction I'd take this in beyond Week 1. Finally, I
can't generalize past the 32 clients in this sample — my own client-holdout
test showed a hand rule that looked accurate on training clients failed on
one it hadn't seen, a caution against overclaiming generalization even when
a model performs well here.
